1. Load libraries

```python
import pandas as pd
import numpy as np

from sdv.metadata import SingleTableMetadata
from sdv.single_table import GaussianCopulaSynthesizer

print("Libraries loaded successfully.")
```

2. Load the preprocessed Cert dataset

```python
data = pd.read_csv(
    r"C:\Users\Admin\OneDrive - Sol Plaatje University\Documents\cert_preprocessed.csv"
)

print("Dataset loaded successfully.")
print("Shape:", data.shape)

data.head()
```

3. Check missing values

```python
print(data.isnull().sum())
```

4. Convert the timestamp to datetime

```python
data["timestamp"] = pd.to_datetime(data["timestamp"])

print(data.dtypes)
```

5. Creation of metadata

```python
metadata = SingleTableMetadata()

metadata.detect_from_dataframe(data)

print("Metadata created successfully.")
```

6. Building Gaussian Copula Model

```python
gaussian = GaussianCopulaSynthesizer(
    metadata=metadata
)

print("Gaussian Copula model created.")
```

7. Train the Gaussian Copula model created

```python
gaussian.fit(data)

print("Training completed successfully.")
```

8. Generate synthetic data

```python
synthetic_gaussian = gaussian.sample(
    num_rows=len(data)
)

print("Synthetic dataset generated.")
```

9. viewing the dataset

```python
synthetic_gaussian.head()
```

10. shape comparison

```python
print("Original Shape :", data.shape)
print("Synthetic Shape:", synthetic_gaussian.shape)
```

11. Compare the data types

```python
comparison = pd.DataFrame({
    "Original": data.dtypes,
    "Synthetic": synthetic_gaussian.dtypes
})

comparison
```

12. Save the synthetic dataset

```python
output_path = r"C:\Users\Admin\OneDrive - Sol Plaatje University\Documents\synthetic_gaussian.csv"

synthetic_gaussian.to_csv(output_path, index=False)

print("Synthetic dataset saved successfully.")
print(output_path)
```
