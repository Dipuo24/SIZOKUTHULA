6. Creation ofthr DBE employee information

```python
import numpy as np

np.random.seed(42)

exam_roles = [
    "Chief Examination Officer",
    "Senior Examination Officer",
    "Provincial Examination Coordinator",
    "Printing Technician",
    "Warehouse Officer",
    "ICT Administrator",
    "Data Capturer"
]

dbe_departments = [
    "Question Paper Development",
    "Secure Printing",
    "Logistics",
    "Results Processing",
    "Provincial Office",
    "Examination Administration"
]

provinces = [
    "Gauteng",
    "Limpopo",
    "Mpumalanga",
    "Free State",
    "North West",
    "KwaZulu-Natal",
    "Western Cape",
    "Eastern Cape",
    "Northern Cape"
]

subjects = [
    "Mathematics",
    "Mathematical Literacy",
    "Physical Sciences",
    "Life Sciences",
    "Accounting",
    "English FAL",
    "English HL",
    "Business Studies",
    "Geography"
]

employees["role"] = np.random.choice(exam_roles, len(employees))

employees["province"] = np.random.choice(provinces, len(employees))

employees["department"] = np.random.choice(
    dbe_departments,
    len(employees)
)

employees["subject_permission"] = np.random.choice(
    subjects,
    len(employees)
)

employees.head()
```

```python
duplicates = employees["user_id"].duplicated().sum()

print("Duplicate user IDs:", duplicates)
```

7. Loading the CERT activity logs

```python
import pandas as pd

ROWS = 10000      # You can increase later to 50000 or 100000

logon = pd.read_csv(logon_file, nrows=ROWS)
device = pd.read_csv(device_file, nrows=ROWS)
file_access = pd.read_csv(file_file, nrows=ROWS)
email = pd.read_csv(email_file, nrows=ROWS)
http = pd.read_csv(http_file, nrows=ROWS)

print("Datasets loaded successfully!")

print(logon.shape)
print(device.shape)
print(file_access.shape)
print(email.shape)
print(http.shape)
```

8. Making all files to have the same columns

```python
logon = logon.rename(columns={
    "date":"timestamp",
    "user":"user_id",
    "pc":"location"
})

device = device.rename(columns={
    "date":"timestamp",
    "user":"user_id",
    "pc":"location"
})

file_access = file_access.rename(columns={
    "date":"timestamp",
    "user":"user_id",
    "pc":"location"
})

email = email.rename(columns={
    "date":"timestamp",
    "user":"user_id",
    "pc":"location",
    "to":"to_email",
    "from":"from_email"
})

http = http.rename(columns={
    "date":"timestamp",
    "user":"user_id",
    "pc":"location"
})
```

9. Creating one activity column

```python
logon["action"] = "Logon"

device["action"] = "USB Device"

file_access["action"] = "File Access"

email["action"] = "Email"

http["action"] = "Web Browsing"
```

10. creating the missing columns

```python
logon["filename"] = None
logon["url"] = None
logon["to_email"] = None
logon["from_email"] = None

device["filename"] = None
device["url"] = None
device["to_email"] = None
device["from_email"] = None

file_access["url"] = None
file_access["to_email"] = None
file_access["from_email"] = None

email["filename"] = None
email["url"] = None

http["filename"] = None
http["to_email"] = None
http["from_email"] = None
```

11. Keeping all useful columns

```python
columns = [
    "timestamp",
    "user_id",
    "location",
    "action",
    "filename",
    "url",
    "to_email",
    "from_email"
]

logon = logon[columns]

device = device[columns]

file_access = file_access[columns]

email = email[columns]

http = http[columns]
```

```python
print(logon.shape)
print(device.shape)
print(file_access.shape)
print(email.shape)
print(http.shape)
```

12. Combining all the files

```python
activities = pd.concat(
    [logon, device, file_access, email, http],
    ignore_index=True
)

print(activities.shape)

activities.head()
```

13. Combining the activity data frame with the employee

```python
activities = activities.merge(
    employees,
    on="user_id",
    how="left"
)

print(activities.shape)

activities.head()
```
