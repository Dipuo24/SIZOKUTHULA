```python
import pandas as pd
import numpy as np
import random

from pathlib import Path
print("Libraries loaded successfully.")
```

Creating behavioural features

```python
activities = pd.read_csv("DBE_Employee_Activities.csv")

activities["timestamp"] = pd.to_datetime(
    activities["timestamp"],
    format="%m/%d/%Y %H:%M:%S"
)

activities["hour"] = activities["timestamp"].dt.hour

activities["day"] = activities["timestamp"].dt.day

activities["month"] = activities["timestamp"].dt.month

activities["weekday"] = activities["timestamp"].dt.day_name()
```

1. DBE employee's after hours activities

```python
activities["after_hours"] = (
    (activities["hour"] < 7) |
    (activities["hour"] > 17)
).astype(int)
```

2. weekend activities

```python
activities["weekend"] = (
    activities["weekday"].isin(
        ["Saturday", "Sunday"]
    )
).astype(int)
```

3. NSC examination months

```python
activities["exam_period"] = (
    activities["month"].isin([10, 11])
).astype(int)
```

4. simulation of records accessed

```python
np.random.seed(42)

activities["records_accessed"] = np.random.poisson(
    lam=8,
    size=len(activities)
)
```

5. simulation of dayas until the examinations

```python
activities["days_to_exam"] = np.random.randint(
    1,
    120,
    len(activities)
)
```
